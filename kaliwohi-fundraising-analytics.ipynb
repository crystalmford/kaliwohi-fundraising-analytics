# Kaliwohi Fundraising Analytics Project

## Project Overview

The Kaliwohi project was a volunteer initiative created with members of a Deloitte and Salesforce Talent Alliance cohort. The goal was to demonstrate how Salesforce and Tableau could support fundraising and donor management for a tribal nonprofit organization.

Because the organization serves survivors of domestic violence, privacy was a critical concern. To protect confidentiality, we used synthetic donor data rather than actual donor records.

This project showcases how Tableau and Salesforce can help nonprofit organizations better understand donor demographics, tribal affiliations, and fundraising opportunities.

---

## What is Kaliwohi?

Kaliwohi is a Cherokee word that means **"integrity."** The name reflects the project's emphasis on trust, confidentiality, and the responsible use of data.

The project was designed as a proof-of-concept analytics solution to demonstrate how data can strengthen nonprofit fundraising efforts while protecting the privacy and dignity of the people being served.

---

## What is One Fire?

One Fire Victim Services is part of the Cherokee Nation and provides advocacy, resources, and support to individuals and families affected by domestic violence.

Because the organization works with highly sensitive information, confidentiality and ethical data practices were central considerations throughout the project.

---

## Project Objectives

- Track donations and donor engagement
- Analyze fundraising performance
- Identify trends across donor demographics and tribal affiliations
- Support strategic fundraising decisions
- Demonstrate how Salesforce can be used to manage donor relationships
- Maintain strict privacy protections

---

## Why Synthetic Data Was Used

The organization serves individuals and families facing domestic violence, making privacy and confidentiality essential.

To protect sensitive information, we created realistic synthetic donor and donation data rather than using actual records. This allowed us to demonstrate Salesforce and Tableau capabilities while ensuring that no real donor or client information was exposed.

By combining privacy-preserving synthetic data with interactive dashboards, the Kaliwohi project shows how analytics can support meaningful work in a safe and ethical way.

## Dashboard Overview

The Tableau dashboard provides an interactive view of fundraising performance, donor demographics, and tribal engagement.

The dashboard was designed to help nonprofit stakeholders quickly answer important questions such as:

- How much money has been raised?
- Which donor groups contribute the most?
- Which tribal communities are most engaged?
- Where are donors located?
- Who are the top contributors?

Interactive filters allow users to explore the data by tribe, age group, gender, and campaign, making the dashboard a flexible decision-making tool for fundraising strategy.

## Dashboard Walkthrough

### Tribal Members by Tribe

This pie chart displays the distribution of tribal affiliations represented in the synthetic donor dataset.

To improve readability, the visualization highlights the ten largest tribal groups and combines the remaining affiliations into an "Other" category. This approach reduces clutter while preserving the overall distribution.

This chart helps stakeholders understand which tribal communities are most represented among donors and can inform outreach and engagement strategies.

---

### Donations by Tribe

This treemap shows the total amount donated by each tribal affiliation.

Larger rectangles represent tribes that contributed more to the fundraising campaign, while color intensity provides an additional visual cue to distinguish categories.

This visualization helps identify which communities generated the highest donation totals and can support strategic planning for future fundraising efforts.

---

### Top Donors by Tribe

This table lists the highest-value donors and their associated tribal affiliations.

The view allows stakeholders to quickly identify major contributors and better understand donor patterns across tribal communities.

This information can support donor stewardship, recognition efforts, and relationship-building strategies.

---

## Interactive Features

The dashboard is fully interactive. Users can click on any tribe in the pie chart, treemap, or donor table to filter the other visualizations automatically.

For example, selecting a specific tribe updates the dashboard to show the number of members associated with that tribe, the total donations contributed by that group, and the top donors within that community.

This makes it easy to answer targeted questions and compare fundraising patterns across different tribal affiliations without needing to modify the underlying data.

---

## Key Insights

Because this project uses synthetic data, the primary goal was to demonstrate how Tableau and Salesforce can be used to organize and visualize fundraising information rather than to draw conclusions from actual donor behavior.

The dashboard shows how nonprofit organizations can:

- Identify which tribal communities contribute the most donations
- Recognize top donors and prioritize relationship-building efforts
- Compare donor participation across tribal affiliations
- Interact with the data to answer specific fundraising questions quickly

In a real-world setting, these insights could help guide outreach strategies, donor recognition, and future fundraising planning.

## Recommendations

- Focus outreach efforts on the tribal communities that show the strongest engagement.

- Use the dashboard to better understand donor interests and tailor future fundraising campaigns.
- Continue using Salesforce and Tableau to organize information and support informed fundraising decisions.

## Personal Reflection


This project was especially meaningful to me because it was based on a real fundraising initiative supporting One Fire Victim Services, a program that assists survivors of domestic violence.

Although the actual fundraising event was much smaller than the synthetic data used in this project, our goal was to demonstrate what could be possible if Salesforce and Tableau were used to organize donor information and analyze fundraising trends at a larger scale.

One of the most important aspects of the project was balancing the need for better reporting with the need to protect highly sensitive information. During the project, I learned that some records were still maintained in locked notebooks because confidentiality was such a critical concern. This reinforced how important privacy and trust are when working with organizations that serve vulnerable communities.

By using realistic synthetic data, our team was able to show how modern analytics tools could help One Fire better understand donor engagement and fundraising opportunities without exposing any confidential information.

This project combined technical skills with a mission that had real human impact. It demonstrated how data can be used not just to answer business questions, but to support organizations doing deeply meaningful work.
